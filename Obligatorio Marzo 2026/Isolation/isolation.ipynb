## Pruebas Generales

```python
agents = ["RandomAgent", "Stratagem", "MinimaxAgent", "ExpectimaxAgent"]
partidos = 200
partidos_totales_agente = partidos * 8
stats = {}

for agent in agents:
    stats[agent] = {
        "victorias_totales": 0,
        "victorias_como_player1": 0,
        "victorias_como_player2": 0,
    }

for agent1 in agents:
    for agent2 in agents:
        victorias_player1 = 0
        for j in range(partidos):
            i = play_vs_other_agent(env, agent1=eval(agent1)(1), agent2=eval(agent2)(2))
            if i == 1:
                stats[agent1]["victorias_totales"] += 1
                stats[agent1]["victorias_como_player1"] += 1
                victorias_player1 += 1
            else:
                stats[agent2]["victorias_totales"] += 1
                stats[agent2]["victorias_como_player2"] += 1

        plt.bar([agent1, agent2], [victorias_player1, partidos - victorias_player1], color=["mediumblue", "limegreen"])
        plt.title(f"{agent1} vs {agent2}")
        plt.ylabel("Victorias")
        plt.show()

plt.bar(stats.keys(), [stats[agent]["victorias_totales"] for agent in agents], color=["yellow", "deepskyblue", "red", "darkviolet"])
plt.title("Victorias totales por agente")
plt.xlabel("Agentes")
plt.ylabel("Victorias")
plt.show()

for agent in agents:
    plt.bar(["Player 1", "Player 2"], [stats[agent]["victorias_como_player1"], stats[agent]["victorias_como_player2"]], color=["darkblue", "green"])
    plt.title(f"Victorias de {agent} como Player 1 y Player 2")
    plt.ylabel("Victorias")
    plt.show()
    print(f"{agent} victorias totales: {stats[agent]['victorias_totales']} victorias")
    print(f"{agent} como Player 1: {stats[agent]['victorias_como_player1']} victorias")
    print(f"{agent} como Player 2: {stats[agent]['victorias_como_player2']} victorias")

victorias_totales_players1 = stats[agents[0]]["victorias_como_player1"] + stats[agents[1]]["victorias_como_player1"] + stats[agents[2]]["victorias_como_player1"] + stats[agents[3]]["victorias_como_player1"]
victorias_totales_players2 = stats[agents[0]]["victorias_como_player2"] + stats[agents[1]]["victorias_como_player2"] + stats[agents[2]]["victorias_como_player2"] + stats[agents[3]]["victorias_como_player2"]
plt.bar(["Player 1", "Player 2"], [victorias_totales_players1, victorias_totales_players2], color=["mediumblue", "limegreen"])
plt.title("Victorias totales en general como Player 1 y Player 2")
plt.ylabel("Victorias")
plt.show()
```

Probando profundidad optima de Minimax

```python
#Se borro sin querer en algun commit, vuelvo a ponerlo a modo ilustrativo pero es por eso que cambian los colores

agents = ["RandomAgent", "Stratagem", "ExpectimaxAgent"]
partidos = 20
profundidades = range (1, 5)
stats = {}

for i in profundidades:
    stats[i] = {
        "victorias_totales": 0,
        "victorias_como_player1": 0,
        "victorias_como_player2": 0,
        "tiempo": 0
    }

for k in profundidades:
    inicio = time.time()
    for agent2 in agents:
        for j in range(partidos):
            i = play_vs_other_agent(env, agent1=MinimaxAgent(1, k), agent2=eval(agent2)(2))
            if i == 1:
                stats[k]["victorias_totales"] += 1
                stats[k]["victorias_como_player1"] += 1

        for j in range(partidos):
            i = play_vs_other_agent(env, agent1=eval(agent2)(1), agent2=MinimaxAgent(2, k))
            if i == 2:
                stats[k]["victorias_totales"] += 1
                stats[k]["victorias_como_player2"] += 1

    fin = time.time()
    stats[k]["tiempo"] = fin - inicio    

for k in profundidades:
    plt.bar(["Player 1", "Player 2"], [stats[k]["victorias_como_player1"], stats[k]["victorias_como_player2"]], color=["darkblue", "green"])
    plt.title(f"Victorias de MinimaxAgent con profundidad {k} como Player 1 y Player 2")
    plt.ylabel("Victorias")
    plt.show()
    print ("Victorias totales con profundidad ", k, ":", stats[k]["victorias_totales"])
    print ("Victorias como player 1 con profundidad", k , ":", stats[k]["victorias_como_player1"])
    print ("Victorias como player 2 con profundidad", k , ":", stats[k]["victorias_como_player2"])
    print ("Tiempo de ejecución con profundidad", k , ":", stats[k]["tiempo"], "segundos")


plt.bar(stats.keys(), [stats[k]["victorias_totales"] for k in profundidades], color=["violet", "deeppink", "darkviolet", "indigo"])
plt.title("Victorias totales por profundidad")
plt.xlabel("Profundidad")
plt.ylabel("Victorias")
plt.show()

plt.bar(stats.keys(), [stats[k]["tiempo"] for k in profundidades], color=["violet", "deeppink", "darkviolet", "indigo"])
plt.title("Tiempo de ejecución por profundidad")
plt.xlabel("Profundidad")
plt.ylabel("Tiempo (segundos)")
plt.show()

plt.scatter([stats[k]["tiempo"] for k in profundidades], [stats[k]["victorias_totales"] for k in profundidades], color=["violet", "deeppink", "darkviolet", "indigo"])
plt.title("Victorias totales en función del tiempo de ejecución")
plt.xlabel("Tiempo (segundos)")
plt.ylabel("Victorias")
plt.show()
```

Prueba de distintas métricas para Minimax

```python
heuristicas = ["baseline", "offensive", "defensive", "simple", "d2o", "o2d"]

partidos = 200
stats = {}

for n in heuristicas:
    stats[n] = {
        "victorias_totales": 0,
        "victorias_como_player1": 0,
        "victorias_como_player2": 0
    }

for n in heuristicas:
    victorias_player1 = 0
    for j in range(partidos):
        i = play_vs_other_agent(env, agent1=MinimaxAgent(1, heuristic=n), agent2=Stratagem(2))
        if i == 1:
            stats[n]["victorias_totales"] += 1
            stats[n]["victorias_como_player1"] += 1
            victorias_player1 += 1

    plt.bar([n, "Stratagem"], [victorias_player1, partidos - victorias_player1], color=["cyan", "lawngreen"])
    plt.title(f"{n} vs Stratagem")
    plt.ylabel("Victorias")
    plt.show()
    print(f"Victorias para {n}: {victorias_player1}, Victorias para Stratagem: {partidos - victorias_player1}")

    victorias_player2 = 0
    for j in range(partidos):
        i = play_vs_other_agent(env, agent1=Stratagem(1), agent2=MinimaxAgent(2, heuristic=n))
        if i == 2:
            stats[n]["victorias_totales"] += 1
            stats[n]["victorias_como_player2"] += 1
            victorias_player2 += 1

    plt.bar(["Stratagem", n], [partidos - victorias_player2, victorias_player2], color=["lawngreen", "cyan"])
    plt.title(f"Stratagem vs {n}")
    plt.ylabel("Victorias")
    plt.show()
    print(f"Victorias para Stratagem: {partidos - victorias_player2}, Victorias para {n}: {victorias_player2}")

for n in heuristicas:
    plt.bar(["Player 1", "Player 2"], [stats[n]["victorias_como_player1"], stats[n]["victorias_como_player2"]], color=["turquoise", "darkgreen"])
    plt.title(f"Victorias de MinimaxAgent con heuristica {n} como Player 1 y Player 2")
    plt.ylabel("Victorias")
    plt.show()
    print(f"Victorias totales con heuristica {n}: {stats[n]['victorias_totales']}")
    print(f"Victorias como player 1 con heuristica {n}: {stats[n]['victorias_como_player1']}")
    print(f"Victorias como player 2 con heuristica {n}: {stats[n]['victorias_como_player2']}")


plt.bar(stats.keys(), [stats[n]["victorias_totales"] for n in heuristicas], color=["lightyellow", "yellow", "orange", "red", "firebrick", "crimson"])
plt.title("Victorias totales por heuristica")
plt.xlabel("Heuristica")
plt.ylabel("Victorias")
plt.show()
```

Analisis impacto poda Alfa-Beta

```python
victorias_sin_poda = 0
tiempo_sin_poda = 0
victorias_con_poda = 0
tiempo_con_poda = 0
partidos = 100
inicio = time.time()
for i in range(partidos):
    j = play_vs_other_agent(env, agent1=MinimaxSinPodaAgent(1), agent2=RandomAgent(2))
    if j == 1:
        victorias_sin_poda += 1
tiempo_sin_poda = time.time() - inicio

inicio = time.time()
for i in range(partidos):
    j = play_vs_other_agent(env, agent1=MinimaxAgent(1), agent2=RandomAgent(2))
    if j == 1:
        victorias_con_poda += 1
tiempo_con_poda = time.time() - inicio

print(f"Victorias sin poda: {victorias_sin_poda}, Tiempo sin poda: {tiempo_sin_poda} segundos")
print(f"Victorias con poda: {victorias_con_poda}, Tiempo con poda: {tiempo_con_poda} segundos")

plt.bar(["Con Poda", "Sin Poda"], [tiempo_con_poda, tiempo_sin_poda], color=["gold", "royalblue"])
plt.title("Tiempo de Ejecución con Poda vs Sin Poda")
plt.ylabel("Tiempo (segundos)")
plt.show()
```
